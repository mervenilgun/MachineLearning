```python
import numpy as np
```

```python
#first array
arr1 = np.array([[4, 7],[2,6]],
               dtype = np.float64)
```

```python
#second array
arr2 = np.array([[3,6],[2,8]],
                dtype = np.float64)
```

```python
#iki arrayi toplama
Sum = np.add(arr1,arr2)
print("Addition of Two Arrays: ")
print(Sum)
```

```python
#Tüm elemanların toplamı
Sum1 = np.sum(arr1)
print("/nAddition of Array elements: ")
print(Sum1)
```

```python
#Tüm elemanların karekökünü alma
Sqrt = np.sqrt(arr1)
print( "/nSquare root af Array1 elements:")
print(Sqrt)
```

```python
#Transpozunu alma 
Trans_arr = arr1.T
print("\nTranspose of Array:  ")
print(Trans_arr)
```
