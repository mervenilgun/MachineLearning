### Veri Görselleştirme 

```python
import matplotlib.pyplot as plt

df['Yaş'].plot(kind= 'hist' , bins=10, title= 'Yaş Dağılımı')
plt.xlabel('Yaş')
plt.xlabel('Frekans')
plt.show()

df.groupby('Şehir')['Yaş'].mean().plot(kind='bar' , title='Şehirlere göre ortalama yaş')
plt.xlabel('Şehir')
plt.xlabel('Ortalama Yaş')
plt.show()
```

#### DataFrame Birleştirme(Merge)

```python
data1 = { 'Anahtar': ['K1' , 'K2', 'K3'],
         'Değer1': ['1' , '2', '3']}

data2 = { 'Anahtar': ['K1' , 'K2', 'K4'],
         'Değer1': ['4' , '5', '6']}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

birleşik_df = pd.merge(df1, df2, on= 'Anahtar' , how= 'inner')
print(birleşik_df)
```

#### DataFrame İstifleme (Concat)

```python
df3 = pd.concat([df1, df2], axis=0)
print(df3)  #alt alta ekler

df4 = pd.concat([df1, df2], axis=1)
print(df4) #yan yana ekler
```
