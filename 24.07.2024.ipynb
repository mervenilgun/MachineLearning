### Özellik vektörünü ve hedef değişkeni belirle

```python
X=  df.drop(['class'], axis=1)
y = df['class']
```

### Eğitim ve Test Kümesi Ayır

```python
from sklearn.model_selection import train_test_split
#print(X.isnull().sum().sum())
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.33, random_state = 42)
```

```python
# X_train and X_test örnek sayısı
X_train.shape, X_test.shape 
print(type (X_train))
```

###  Öznitelik Mühendisliği
Özellik Mühendisliği, ham verileri modelimizi daha iyi anlamamıza ve tahmin gücünü artırmamıza yardımcı olan yararlı özelliklere dönüştürme sürecidir. Özellik mühendisliğini farklı değişken türleri üzerinde gerçekleştireceğim.

```python
# check data types in X_train
X_train.dtypes
print(type(X_train))
```

```python
X_train.head()
```

We can see that all the variables are ordinal categorical data type

```python
#import category encoders
from sklearn.preprocessing import OrdinalEncoder
```

```python
# encode categorical variables with ordinal encoding
encoder = OrdinalEncoder()
X_train= pd.DataFrame(encoder.fit_transform(np.array(X_train)))
X_test = pd.DataFrame(encoder.fit_transform(X_test))
```

```python
X_train.head()
```

###  Varsayılan parametrelerle Rastgele Orman Sınıflandırıcı modeli

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier (random_state=0)

rfc.fit(X_train, y_train)

y_pred = rfc.predict(X_test)

from sklearn.metrics import accuracy_score

print('Modelin 10 karar ağacıyla doğruluk değeri: (0:0.4f)'. format(accuracy_score(y_test, y_pred)))
```

###  n_estimators=100 parametresi ile Rastgele Orman Sınıflandırıcı modeli

```python
rfc_100 Random ForestClassifier(n_estimators=100, random_state=0)
rfc_100.fit(X_train, y_train)
y_pred_100 rfc_100.predict(X_test)
print('Modelin 100 karar ağacıyla doğruluk değeri:: {0:0.4f)'. format(accuracy_score(y_test, y_pred_100)))
```
