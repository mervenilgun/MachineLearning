```python
year = [ 1972, 1982, 1992, 2002, 2012]
e_india = [100.6, 158.61, 305.54, 394.56, 724.76]
e_bangladesh = [10.5, 25.21, 58.55, 119.27, 274.87]
```

### çizgi stilinin biçimlendirilmesi ve koordinatların çizilmesi 

```python
plt.plot(year, e_india, color = 'orange',
        marker = 'o', markersize = 12,
        label = 'India')
plt.plot(year, e_bangladesh, color = 'g',
        linestyle = 'dashed', linewidth = 2,
        label = 'Bangladesh')
plt.xlabel('Years')
plt.ylabel('Power consumption in kWh')

plt.title('Electricity consumption per \ capita of India and Bangladesh')

plt.legend()
plt.show()
```

## Gruplandırılmış ve Yığılmış Çubuk Grafikleri

```python
import numpy as np
import matplotlib.pyplot as plt

categories = ['A', 'B', 'C', 'D']
values1 = [3, 7, 5, 12]
values2 = [4, 6, 8, 10]
values3 = [8, 9, 3, 7]

x = np.arange(len(categories))
print(x)

#plt.bar(x, height, width, bottom, align)
plt.bar(x - 0.2, values1, width= 0.2, color= "pink", label='Set 1')
plt.bar(x      , values2, width= 0.2, color= "purple", label='Set 2')
plt.bar(x + 0.2, values3, width= 0.2, color= "lightblue", label='Set 3')

plt.xticks(x, categories)
plt.title('Gruplandırılmış Çubuk Grafiği')
plt.xlabel('Kategoriler')
plt.ylabel('Değerler')
plt.legend()

plt.show()
```

```python
plt.bar(categories, values1, color = "red", label='Set 1')
plt.bar(categories, values2, color= "lightblue", bottom=values1, label= 'Set2')

plt.title('Gruplandırılmış Çubuk Grafiği')
plt.xlabel('Kategoriler')
plt.ylabel('Değerler')
plt.legend() #köşeye set1 set2 yazdırır

plt.show()
```

## Pasta Grafikleri (Pie Charts) ##

```python
import matplotlib.pyplot as plt
 
#veriseti
labels= ['Elma', 'Nar', 'Kiraz', 'Portakal']
sizes = [15, 30, 45, 100]

#pasta grafiği oluşturma
plt.pie(sizes, labels=labels, autopct= '%1.1f%%')

#başlık
plt.title('Meyve Dağılımı')

#Gösterim
plt.show()
```

## Histogramlar ##

```python
import matplotlib.pyplot as plt
import numpy as np

# veri seti
data = np.random.randn(1000)

#histogram oluşturma
plt.hist(data, bins=10, histtype='stepfilled',) #histype='bar'

#başlık ve etiketler
plt.title('Histogram')
plt.xlabel('Değerler')
plt.ylabel('Frekans')

#Gösterim
plt.show()
```

## Dağılım Grafikleri (Scatter Plots) ##

```python
# iki değişken arasındaki ilişkiyi gösterir
import matplotlib.pyplot as plt

#veri seti
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 35]

#Dağılım grafği oluşturması
plt.scatter(x, y, marker= '*', color='red', linewidths=12, edgecolor='red')

#başlık ve etiketler
plt.title('Dağılım Grafiği')
plt.xlabel(' X Değeri')
plt.ylabel('Y Değeri')

#Gösterim
plt.show()
```

### Renk ve Boyut Özelleştirmesi

```python
colors = [10, 20, 30, 40, 50]
sizes = [100, 200, 300, 400, 500]

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='viridis')
plt.colorbar() #renk skalası ekler

#başlık ve etiketler
plt.title('Dağılım Grafiği')
plt.xlabel(' X Değeri')
plt.ylabel('Y Değeri')

#Gösterim
plt.show()
```

### Grafik Özelleştirme

Renkler ve Çizgi Stilleri

```python
plt.plot(x, y, color='red', linestyle='--', linewidth=2)
plt.show()
```

### Grafik Boyutları ve Çözünürlük

```python
plt.figure(figsize=(10, 6), dpi= 100)
plt.plot(x,y)
plt.show()
```

## Alt Grafikler(Subplots)

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2,2)
fig.suptitle('Dört şekil için ortak başlık')

axs[0, 0].plot(x,y)
axs[0, 0].set_title('Grafik 1')

axs[0, 1].bar(categories, values)
axs[0, 1].set_title('Grafik 2')

axs[1, 0].hist(data)
axs[1, 0].set_title('Grafik 3')

axs[1, 1].pie(sizes, labels=sizes, autopct='%1.1f%%')
axs[1, 1].set_title('Grafik 4')

plt.tight_layout()
plt.show()
```
