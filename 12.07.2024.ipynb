```python
df['Price'] = df['Price'].str.replace('$', "", regex=True).str.replace(",","").astype(float).astype(int)
```

```python
#car_sales['Price']=car_sales['Price'].str.replace('$', '',regex=False).str.replace(',', '',regex=False).astype(float).astype(int)
```

```python
#satış tarihi ekleme
df["Sale Date"] = pd.date_range("1/1/2024", periods=len(df))
df
```

```python
#Toplam kümülatif satış ekleme
df["Total Sales"] = df['Price'].cumsum()
df
```

```python
#Tarihe göre toplam satış miktarı
import matplotlib.pyplot as plt
df.plot(x = "Sale Date", y = "Total Sales");  # x ve y DataFrame deki etiketler olmalıdır 
```
