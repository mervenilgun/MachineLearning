```python
import numpy as np
def activation(z,kind):
    
    implemented = {'linear', 'unit step','logistic','piecewise linear','tanh'}
    
    if kind not in implemented:
        raise AttributeError('%s not in %s'% (kind,implemented))
    
    if kind == 'unit step':
        return np.where(z>=0.0,1,0)
    
    elif kind =='logistic':
        return 1.0 / (1.0+np.exp(-z))
    
    elif kind=='tanh':
        e_p= np.exp(z)
        e_m=np.exp(z)
        return (e_p - e_m) / (e_p + e_m)
    
    elif kind =='piecewise linear':
        if z >= 0.5:
            return 1
        elif z <= -0.5:
            return 0
        else:
            return z + 0.5
    
    else:
        return z

x = np.arange(-5, 5, 0.005)
for a in ('linear', 'unit step', 'logistic', 'piecewise linear', 'tanh'):
    y = [activation(z, kind=a) for z in x]
    plt.plot(x,y)
    plt.title(a)
    plt.ylim(-1.5,1.5)
    plt.xlabel('z')
    plt.grid()
    plt.show()
```

```python
import pandas as pd
import sklearn
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
```

```python
names = ['sepal-lenght', 'sepal-width', 'petal-lenght', 'petal-width', 'Class']
```

```python
irisdata = pd.read.csv(url, names=names)
```

```python
X = irisdata.iloc[:, 0:4]
y = irisdata.select_dtypes(include=[object])
```

```python
X.head()
```

```python
y.Class.unique()
```

```python
le = preprocessing.LabelEncoder()
y = y.apply(le.fit_transform)
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
mlp = MLPClassifier(hidden_layer_sizes=(20, 20, 20), max_iter=1000)
mlp.fit(X_train, y_train.values.ravel())
```

```python
predictions = mlp.predict(X_test)
```

```python
print(predictions)
```

```python
#last thing: evaluation of algorithm performance in classifying flowers
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
```

```python
cm=confusion_matrix(y_test,predictions)
print(cm)
print(classification_report(y_test,predictions))
```

```python
#create a heatmap
plt.figure:(figsize=(10, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
```

```python
# set axis labels and title
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
```
